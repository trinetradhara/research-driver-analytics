```python
import pandas as pd

# Load the CSV
df = pd.read_csv('/content/dhrusti_out - dhrusti_out.csv')

# Filter rows where right pupil is detected
df = df[df['right_pupil_detected'] == 't']

# Sort by image_timestamp
df = df.sort_values('image_timestamp')

# Create 't' as sequential integers starting from 1
df['t'] = range(1, len(df) + 1)

# Select x, y from right pupil center
df['x'] = df['right_pupil_center_x']
df['y'] = df['right_pupil_center_y']

# Calculate dt (time difference in seconds)
df['dt'] = df['image_timestamp'].diff() / 1_000_000_000  # Convert nanoseconds to seconds
df['dt'].iloc[0] = df['dt'].iloc[1]  # Set first row's dt to second row's dt

# Select only the columns VETA needs
veta_df = df[['t', 'x', 'y', 'dt']]

# Save to a new CSV
veta_df.to_csv('veta_ready.csv', index=False)
print(f"Created veta_ready.csv with {len(veta_df)} rows")
```
